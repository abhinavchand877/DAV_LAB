# Experiment 2: Exploratory Data Analysis (EDA)

## Aim
To load a dataset, compute descriptive statistics, and identify missing values and duplicate records.

## Dataset
`student_performance.csv` - 205 records of student academic details (study hours, attendance, GPA, internal and final marks).

## Tools Used
Python 3, Pandas, NumPy

## 1. Importing Libraries

```python
import numpy as np
import pandas as pd
```

## 2. Loading the Dataset

```python
df = pd.read_csv("student_performance.csv")
print("Dataset loaded successfully")
print("Rows, Columns:", df.shape)
```

```python
df.tail()
```

## 3. Understanding the Structure

```python
print("Columns:", list(df.columns))
print()
df.info()
```

```python
df.dtypes
```

## 4. Descriptive Statistics

### 4.1 Numerical columns

```python
df.describe().round(2)
```

```python
num_cols = df.select_dtypes(include="number").columns
stats = pd.DataFrame({
    "Mean":     df[num_cols].mean(),
    "Median":   df[num_cols].median(),
    "Std Dev":  df[num_cols].std(),
    "Variance": df[num_cols].var(),
    "Min":      df[num_cols].min(),
    "Max":      df[num_cols].max(),
    "Skewness": df[num_cols].skew(),
}).round(2)
stats
```

```python
print("Range of Final_Marks :", df["Final_Marks"].max() - df["Final_Marks"].min())
print("Mode of Study_Hours  :", df["Study_Hours"].mode()[0])
print("Quartiles of Final_Marks:")
df["Final_Marks"].quantile([0.25, 0.5, 0.75])
```

### 4.2 Categorical columns

```python
df.describe(include="object")
```

```python
print("Students per Department:")
print(df["Department"].value_counts(), "\n")
print("Students per Gender:")
print(df["Gender"].value_counts())
```

```python
# Average performance by department
df.groupby("Department")[["Study_Hours", "Attendance_Percent", "Final_Marks"]].mean().round(2)
```

## 5. Identifying Missing Values

```python
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
pd.DataFrame({"Missing Count": missing, "Missing %": missing_pct})
```

```python
print("Total missing values:", df.isnull().sum().sum())
print("Rows with at least one missing value:", df.isnull().any(axis=1).sum())
print("Complete rows:", df.notnull().all(axis=1).sum())
```

```python
# Rows that contain missing values
df[df.isnull().any(axis=1)].head(10)
```

## 6. Identifying Duplicate Records

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Show every copy of a duplicated record
df[df.duplicated(keep=False)].sort_values("Student_ID")
```

```python
# Duplicates on the key column
print("Duplicate Student_IDs:", df["Student_ID"].duplicated().sum())

# Removing the duplicates (keeps the first occurrence)
df_clean = df.drop_duplicates()
print("Shape before:", df.shape)
print("Shape after :", df_clean.shape)
```

## 7. Summary of Findings

```python
print("EDA SUMMARY")
print("-" * 40)
print("Total records           :", df.shape[0])
print("Total features          :", df.shape[1])
print("Numerical features      :", len(num_cols))
print("Categorical features    :", df.select_dtypes(include='object').shape[1])
print("Missing values (total)  :", df.isnull().sum().sum())
print("Columns with missing    :", list(df.columns[df.isnull().any()]))
print("Duplicate rows          :", df.duplicated().sum())
```

## Result
The dataset was loaded and inspected. Descriptive statistics were computed for numerical and categorical columns, and missing values and duplicate rows were identified and quantified.

## Conclusion
EDA shows that the data contains missing entries in several columns and a few duplicate records. These must be handled (imputation, removal) before modelling, which is covered in the following experiments.
