```python
import pandas as pd

cafe = pd.read_csv("cafe_sales.csv")

print("Rows x Columns:", cafe.shape)
print("Column names  :", list(cafe.columns))
print(cafe.head(8))
print()
cafe.info()
```

```python
print("Summary statistics:\n", cafe.describe().round(2))
print("\nItems per category:\n", cafe["Category"].value_counts())
print("\nAverage price by category:\n", cafe.groupby("Category")["Price"].mean().round(1))
```

```python
miss = cafe.isnull().sum()
print("Missing values per column:\n", miss)
print("\nPercentage missing:\n", (miss / len(cafe) * 100).round(1))
print("\nRows with missing data:\n", cafe[cafe.isnull().any(axis=1)])

print("\nDuplicate rows found:", cafe.duplicated().sum())
print(cafe[cafe.duplicated(keep=False)])
```
