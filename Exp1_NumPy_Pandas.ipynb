```python
import numpy as np
import pandas as pd

# ---- NumPy ----
temps = np.array([28.5, 31.2, 29.8, 33.0, 30.4, 27.9, 32.1])
print("Temperatures:", temps)
print("Average     :", round(temps.mean(), 2))
print("Hottest/Coolest:", temps.max(), "/", temps.min())
print("In Fahrenheit:", temps * 9 / 5 + 32)
print("Days above 30:", temps[temps > 30])
print("Last three days:", temps[-3:])
print("Reversed:", temps[::-1])
```

```python
sales = np.arange(1, 13).reshape(3, 4) * 10
print("Weekly sales (3 shops x 4 weeks):\n", sales)
print("Shop-wise total :", sales.sum(axis=1))
print("Week-wise mean  :", sales.mean(axis=0))
print("Middle row      :", sales[1])
print("Last column     :", sales[:, -1])
print("Sub-block:\n", sales[0:2, 1:3])
print("Transpose shape :", sales.T.shape)
```

```python
# ---- Pandas ----
orders = pd.DataFrame({
    "Customer": ["Meena", "Raj", "Sunil", "Divya", "Karan"],
    "City":     ["Mysuru", "Bengaluru", "Mysuru", "Mandya", "Bengaluru"],
    "Amount":   [450, 820, 230, 610, 975],
    "Items":    [3, 5, 2, 4, 6],
})
print(orders, "\n")

print("Shape:", orders.shape)
print("Row 3 using loc :", orders.loc[3, "Customer"])
print("Rows 1-3 using iloc:\n", orders.iloc[1:4, 0:3])

orders["PerItem"] = (orders["Amount"] / orders["Items"]).round(1)
print("\nSorted by Amount (high to low):\n", orders.sort_values("Amount", ascending=False))
print("\nOrders above 500 from Bengaluru:\n", orders[(orders["Amount"] > 500) & (orders["City"] == "Bengaluru")])
print("\nCity-wise average amount:\n", orders.groupby("City")["Amount"].mean())
```
