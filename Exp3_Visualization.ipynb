```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("darkgrid")

cafe = pd.read_csv("cafe_sales.csv").drop_duplicates().dropna()
print("Rows used:", len(cafe))

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].hist(cafe["Price"], bins=7, color="coral", edgecolor="white")
ax[0].set_title("Distribution of Price")
ax[0].set_xlabel("Price (Rs)")
sns.histplot(cafe["Rating"], bins=6, kde=True, color="slateblue", ax=ax[1])
ax[1].set_title("Distribution of Rating")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
sns.boxplot(x="Category", y="Price", data=cafe, ax=ax[0])
ax[0].set_title("Price by Category")
sns.boxplot(x="Category", y="Quantity", data=cafe, ax=ax[1])
ax[1].set_title("Quantity Sold by Category")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(x="Price", y="Quantity", hue="Category", size="Rating", data=cafe)
plt.title("Price vs Quantity Sold")
plt.show()

sns.pairplot(cafe, hue="Category", height=2)
plt.show()
```

```python
corr = cafe[["Price", "Quantity", "Rating"]].corr()
print(corr.round(2))

plt.figure(figsize=(5, 4))
sns.heatmap(corr, annot=True, cmap="YlGnBu", square=True)
plt.title("Correlation between Price, Quantity and Rating")
plt.show()
```
